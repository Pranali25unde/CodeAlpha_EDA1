```python
import pandas as pd
```

```python
data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010],
    "Product": ["Laptop", "Mouse", "Keyboard", "Laptop", "Monitor",
                "Mouse", "Keyboard", "Laptop", "Monitor", "Mouse"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics",
                 "Electronics", "Accessories", "Accessories", "Electronics",
                 "Electronics", "Accessories"],
    "Quantity": [2, 5, 3, 1, 2, 4, 6, 1, 3, 5],
    "Price": [50000, 800, 1500, 50000, 12000, 800, 1500, 50000, 12000, 800]
}

df = pd.DataFrame(data)

df["Sales"] = df["Quantity"] * df["Price"]

df
```

```python
print("First 5 rows:")
display(df.head())

print("Dataset Shape:")
print(df.shape)

print("Column Names:")
print(df.columns)

print("Dataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python
print("Statistical Summary:")
display(df.describe())
```

```python
print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Total Quantity Sold:", df["Quantity"].sum())
```

```python
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)

print("Sales by Product:")
display(product_sales)
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

print("Sales by Category:")
display(category_sales)
```

```python
product_quantity = df.groupby("Product")["Quantity"].sum().sort_values(ascending=False)

print("Quantity Sold by Product:")
display(product_quantity)
```

```python
import matplotlib.pyplot as plt

product_sales.plot(kind="bar")

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
product_quantity.plot(kind="bar")

plt.title("Quantity Sold by Product")
plt.xlabel("Product")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=0)
plt.show()
```

```python
print("FINAL INSIGHTS")
print("----------------")

top_product = product_sales.idxmax()
top_product_sales = product_sales.max()

top_category = category_sales.idxmax()
top_category_sales = category_sales.max()

top_quantity_product = product_quantity.idxmax()
top_quantity = product_quantity.max()

print(f"1. Highest sales product: {top_product}")
print(f"   Sales: ₹{top_product_sales:,.0f}")

print(f"2. Highest sales category: {top_category}")
print(f"   Sales: ₹{top_category_sales:,.0f}")

print(f"3. Most sold product by quantity: {top_quantity_product}")
print(f"   Quantity sold: {top_quantity}")

print(f"4. Total sales: ₹{df['Sales'].sum():,.0f}")
print(f"5. Average sales per order: ₹{df['Sales'].mean():,.0f}")
```

```python
df.to_csv("sales_data.csv", index=False)

print("CSV file saved successfully!")
```
